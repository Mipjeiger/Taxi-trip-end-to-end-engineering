## Get latitude and longitude column

```python
from geopy.geocoders import Nominatim
import time

geolocator = Nominatim(user_agent="ride_app")

def get_lat_lon(location):
    if not location or pd.isna(location):
        return None, None
    
    try:
        loc = geolocator.geocode(location)
        time.sleep(1)  # To respect the usage policy of Nominatim
        return (loc.latitude, loc.longitude) if loc else (None, None)
    except:
        return None, None
    
# Usage
get_lat_lon("Jakarta, Indonesia")
```

```python
# Get a list of all Unique Locations
unique_locs = pd.concat([df['Pickup Location'], df['Drop Location']]).unique()
print(f"Total unique locations: {len(unique_locs)}")

# Geocode only the unique locations
cache = {}
for loc in unique_locs:
    cache[loc] = get_lat_lon(loc)

# Map the results back to original DataFrame
df['pickup_lat'], df['pickup_lon'] = zip(*df['Pickup Location'].map(cache))
df['drop_lat'], df['drop_lon'] = zip(*df['Drop Location'].map(cache))
```

```python
df['Datetime'] = pd.date_range(start='2026-01-01', periods=len(df), freq='T')
df.head()
```

```python
df['completed_at'] = df['Datetime'] + pd.to_timedelta(df['estimated_drop_time_minute'], unit='m')
df.head()
```
