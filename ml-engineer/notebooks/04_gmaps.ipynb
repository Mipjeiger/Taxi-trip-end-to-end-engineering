## Multiple ride maps

```python
from geopy.geocoders import Nominatim
import time

geolocator = Nominatim(user_agent="ride_app")

def get_lat_lon(location):
    try:
        loc = geolocator.geocode(location)
        time.sleep(1)  # To respect the usage policy of Nominatim
        return loc.latitude, loc.longitude
    except:
        return None, None
```

## Function for distance pickup location - drop location estimation on google maps

```python
df_sample = df.head(100)  # Sample for testing
df_sample["pickup_lat"], df_sample["pickup_lon"] = zip(*df_sample["Pickup Location"].apply(get_lat_lon))
df_sample["drop_lat"], df_sample["drop_lon"] = zip(*df_sample["Drop Location"].apply(get_lat_lon))
```

```python
df_sample[["Pickup Location", "Drop Location"]].head(10)
```

```python
# Get estimation latitude and longitude using geopy (Compare toward to pickup and drop location with value from geopy)
df_compare = df_sample[["pickup_lat", "pickup_lon", "drop_lat", "drop_lon"]].copy()

# Call geopy
def get_estimation_lat_lon(row):
    pickup_location = row["pickup_lat"], row["pickup_lon"]
    drop_location = row["drop_lat"], row["drop_lon"]

    # Get estimation for pickup and drop location using geopy
    pickup_estimation = geolocator.reverse(pickup_location, exactly_one=True)
    drop_estimation = geolocator.reverse(drop_location, exactly_one=True)
    
    return pickup_estimation.latitude, pickup_estimation.longitude, drop_estimation.latitude, drop_estimation.longitude

# Usage
get_estimation_lat_lon(df_compare.iloc[0])
```

## Interactive UI map for (production estimated)

```python
# Interactive Map with Folium (Direct Display)
import folium
from folium import plugins
import pandas as pd
from IPython.display import display, HTML
import webbrowser
import os

# 1.  Data check
required_cols = ["pickup_lat", "pickup_lon", "drop_lat", "drop_lon"]

for col in required_cols:
    if col not in df_sample.columns:
        raise ValueError(f"Missing required column: {col}")
    
df_clean = df_sample.dropna(subset=required_cols).copy()
print(f"✅ Cleaned data: {len(df_clean)} valid routes out of {len(df_sample)} total.")

# 2. Map center
center_lat = df_clean["pickup_lat"].mean()
center_lon = df_clean["pickup_lon"].mean()
print(f"📍 Map center: ({center_lat:.4f}, {center_lon:.4f})")
print(f"📊 Total valid routes: {len(df_clean)}")

# 3. Create folium map
ride_map = folium.Map(
    location=[center_lat, center_lon],
    zoom_start=12,
    tiles='https://{s}.tile.openstreetmap.org/{z}/{x}/{y}.png',
    attr='OpenStreetMap'
)

# 4. Plot all routes
for idx, row in df_clean.head(30).iterrows():
    try:
        # Debug info
        pickup_lat = row["pickup_lat"]
        pickup_lon = row["pickup_lon"]
        drop_lat = row["drop_lat"]
        drop_lon = row["drop_lon"]

        # Pickup marker (green)
        folium.Marker(
            location=[row["pickup_lat"], row["pickup_lon"]],
            popup=f"<b>Pickup {idx+1}</b><br>Lat: {row['pickup_lat']:.4f}<br>Lon: {row['pickup_lon']:.4f}",
            tooltip=f"Pickup {idx+1}",
            icon=folium.Icon(color='green', icon='play', icon_color='white')
        ).add_to(ride_map)
        
        # Drop marker (red)
        folium.Marker(
            location=[row["drop_lat"], row["drop_lon"]],
            popup=f"<b>Drop {idx+1}</b><br>Lat: {row['drop_lat']:.4f}<br>Lon: {row['drop_lon']:.4f}",
            tooltip=f"Drop {idx+1}",
            icon=folium.Icon(color='red', icon='stop', icon_color='white')
        ).add_to(ride_map)
        
        # Draw route line
        folium.PolyLine(
            locations=[
                [row["pickup_lat"], row["pickup_lon"]],
                [row["drop_lat"], row["drop_lon"]]
            ],
            color='blue',
            weight=2,
            opacity=0.7,
            popup=f"Route {idx+1}"
        ).add_to(ride_map)
        
    except Exception as e:
        print(f"❌ Error plotting route {idx}: {e}")

# 5. Add title/legend
legend_html = f"""

<div style="
    position: fixed;
    top: 10px; left: 10px;
    width: 200px;
    background-color: white;
    border:2px solid grey;
    z-index:9999;
    font-size:14px;
    padding:10px;
    border-radius:6px;">
<b>Routes Overview</b><br>
🟢 Pickup<br>
🔴 Drop<br>
🔵 Route<br><br>
<b>Total:</b> {len(df_clean)}
</div>
"""
ride_map.get_root().html.add_child(folium.Element(legend_html))

# 6. Display map
display(HTML(ride_map._repr_html_()))

# 7. Fallback: Save and open in browser
output_file = "../../ml-engineer/database/html/ride_map.html"
ride_map.save(output_file)

print(f"💾 Saved map to: {output_file}")

# Auto open in default browser (optional)
try:
    webbrowser.open(f"file://{os.path.abspath(output_file)}")
except Exception as e:
    print(f"Could not open browser automatically: {e}")
```

## Create route effience

```python
df["Efficiency_routes"] = df["Ride Distance"] / df["Avg CTAT"] # Efficiency = Distance / Time average speed (km/h)

top_routes = (
    df.groupby(["Pickup Location", "Drop Location"])
    .agg({
        "Efficiency_routes": "mean",
        "Avg CTAT": "mean",
        "Ride Distance": "mean",
        "route_count": "mean"
    })
    .sort_values(by="Efficiency_routes", ascending=False).head(20)
)
top_routes
```

```python
# Create plot for efficiency routes (simple approach)
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))
route_names = [f"{pickup} → {drop}" for pickup, drop in top_routes.index]
plt.barh(route_names, top_routes["Efficiency_routes"], color='skyblue')
plt.title("Top 20 Most Efficient Routes (Distance/Time)", fontsize=16)
plt.xlabel("Efficiency (km/h)", fontsize=14)
plt.ylabel("Route (Pickup → Drop)", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Build comprehensive route analysis dashboard
def analyze_route(idx, show_map=True, show_details=True):
    """
    Comprehensive route analyzer with metrics and visualization.
    
    Parameters:
    - idx: route index to analyze
    - show_map: whether to display Google Maps (default: True)
    - show_details: whether to show detailed metrics (default: True)
    """
    
    # Validate index
    if idx not in df.index:
        print(f"❌ Error: Index {idx} not found in dataset. Valid range: 0-{len(df)-1}")
        return None
    
    row = df.loc[idx]
    
    # Create beautiful dashboard output
    print("\n" + "="*70)
    print("🚗 ROUTE ANALYSIS DASHBOARD")
    print("="*70)
    
    # Location Information
    print("\n📍 LOCATION INFORMATION:")
    print(f"  • Pickup:    {row['Pickup Location']}")
    print(f"  • Drop:      {row['Drop Location']}")
    
    # Trip Metrics
    print("\n📊 TRIP METRICS:")
    print(f"  • Distance:  {row['Ride Distance']:.2f} km")
    print(f"  • Time:      {row['Avg CTAT']:.2f} minutes")
    print(f"  • Price:     IDR {row['Booking Value']:.2f}")
    
    # Calculated Metrics
    if row['Avg CTAT'] > 0:
        speed = (row['Ride Distance'] / row['Avg CTAT']) * 60  # km/h
        cost_per_km = row['Booking Value'] / row['Ride Distance'] if row['Ride Distance'] > 0 else 0
        
        print("\n⚡ EFFICIENCY METRICS:")
        print(f"  • Average Speed:    {speed:.2f} km/h")
        print(f"  • Cost per km:      IDR {cost_per_km:.2f}")
    
    # Additional Information
    if 'Traffic Status' in row:
        print(f"\n🚦 TRAFFIC STATUS: {row['Traffic Status']}")
    
    print("\n" + "="*70 + "\n")
    
    # Display map if requested
    if show_map:
        return show_route(df, idx)

# Interactive route selector
def show_route_summary(start_idx=0, num_routes=5):
    """Show summary of multiple routes"""
    print("\n" + "="*70)
    print(f"📋 ROUTE SUMMARY - Routes {start_idx} to {start_idx + num_routes - 1}")
    print("="*70 + "\n")
    
    for i in range(start_idx, min(start_idx + num_routes, len(df))):
        row = df.loc[i]
        print(f"Route {i}: {row['Pickup Location']:25} → {row['Drop Location']:25} | "
              f"Distance: {row['Ride Distance']:6.2f} km | "
              f"Time: {row['Avg CTAT']:6.2f} min | "
              f"Price: IDR {row['Booking Value']:8.2f}")
    
    print("="*70 + "\n")

# Example usage - Analyze individual route
print("✅ Route Analyzer Ready!\n")
analyze_route(idx=1, show_map=True, show_details=True)

# Optional: Show summary of multiple routes
# show_route_summary(start_idx=0, num_routes=5)
```
